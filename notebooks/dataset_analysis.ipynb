# 20 Newsgroups Dataset Exploration & Analysis

This notebook provides reproducible exploration of the official **20 Newsgroups** corpus retrieved via `sklearn.datasets.fetch_20newsgroups`.

```python
import os
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_20newsgroups

# Add project root to sys.path
BASE_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(BASE_DIR) not in sys.path:
    sys.path.append(str(BASE_DIR))

from core.config import DOCUMENTS_CSV, METADATA_CSV
```

## 1. Fetching Dataset & Loading Data

```python
# Load pre-saved documents CSV or fetch directly
if DOCUMENTS_CSV.exists():
    df = pd.read_csv(DOCUMENTS_CSV)
else:
    raw = fetch_20newsgroups(subset='all', remove=(), shuffle=False)
    df = pd.DataFrame({
        'doc_id': [f'doc_{i+1:06d}' for i in range(len(raw.data))],
        'category_id': raw.target,
        'category_name': [raw.target_names[i] for i in raw.target],
        'raw_text': raw.data
    })

print(f"Total Documents: {len(df)}")
print(f"Total Categories: {df['category_name'].nunique()}")
df.head()
```

## 2. Category Distribution

```python
plt.figure(figsize=(12, 6))
cat_counts = df['category_name'].value_counts()
sns.barplot(x=cat_counts.values, y=cat_counts.index, palette='viridis')
plt.title('20 Newsgroups Document Distribution per Category', fontsize=14)
plt.xlabel('Number of Documents')
plt.ylabel('Category Name')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## 3. Document Length Summary

```python
df['char_length'] = df['raw_text'].apply(len)
df['word_count'] = df['raw_text'].apply(lambda x: len(x.split()))

df[['char_length', 'word_count']].describe()
```
