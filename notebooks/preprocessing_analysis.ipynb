# Preprocessing Pipeline Analysis & Verification

This notebook demonstrates the text preprocessing pipeline applied to the **20 Newsgroups** corpus.

```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

BASE_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(BASE_DIR) not in sys.path:
    sys.path.append(str(BASE_DIR))

from core.config import DOCUMENTS_CSV
from preprocessing.preprocess import preprocess_text, preprocess_debug
```

## 1. Load Dataset

```python
df = pd.read_csv(DOCUMENTS_CSV)
print(f"Loaded {len(df)} documents.")
df.head(3)
```

## 2. Before / After Preprocessing Statistics

```python
df['cleaned_text'] = df['raw_text'].fillna('').apply(preprocess_text)
df['raw_char_len'] = df['raw_text'].fillna('').apply(len)
df['cleaned_char_len'] = df['cleaned_text'].apply(len)
df['raw_token_count'] = df['raw_text'].fillna('').apply(lambda x: len(x.split()))
df['cleaned_token_count'] = df['cleaned_text'].apply(lambda x: len(x.split()))

summary = df[['raw_char_len', 'cleaned_char_len', 'raw_token_count', 'cleaned_token_count']].describe()
summary
```

## 3. Sample Debug Transformation Across Categories

```python
categories = df['category_name'].unique()[:4]
for cat in categories:
    sample_doc = df[df['category_name'] == cat].iloc[0]
    debug_res = preprocess_debug(sample_doc['raw_text'])
    print(f"=== Category: {cat} (Doc ID: {sample_doc['doc_id']}) ===")
    print("--- RAW TEXT (First 150 chars) ---")
    print(debug_res['raw_text'][:150] + '...')
    print("--- CLEANED & STEMMED TEXT ---")
    print(debug_res['final_cleaned'][:150] + '...')
    print('\n' + '='*60 + '\n')
```
